```python
import itertools

def evaluar_clausula(clausula, asignacion):
    for literal in clausula:
        var = abs(literal)
        if (literal > 0 and asignacion[var] == 1) or (literal < 0 and asignacion[var] == 0):
            return True
    return False

def sat_bruteforce(formula, n):
    for valores in itertools.product([0, 1], repeat=n):
        asignacion = {i+1: valores[i] for i in range(n)}

        if all(evaluar_clausula(c, asignacion) for c in formula):
            print("Solución encontrada:", asignacion)
            return True

    return False


# Ejemplo
formula = [
    [1, -2, 3],
    [-1, 2, 4]
]

n = 4

if sat_bruteforce(formula, n):
    print("La fórmula es SATISFACIBLE")
else:
    print("La fórmula NO es satisfacible")
```
